```python
sales_data_orders.groupby("channel").size()
```

```python
revenue_by_channel=(sales_data_orders
                    .groupby("channel")["revenue"]
                    .sum()
                    .reset_index(name="total_revenue"))

revenue_by_channel
```

```python
revenue_by_channel_average=(sales_data_orders
                    .groupby("channel")["revenue"]
                    .mean()
                    .reset_index(name="average_revenue"))

revenue_by_channel_average
```

```python
quantity_by_channel=(sales_data_orders
                    .groupby("channel")["quantity"]
                    .sum()
                    .reset_index(name="total_quantity"))

quantity_by_channel
```
