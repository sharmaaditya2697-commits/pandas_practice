```python
#sales_data_orders.groupby("channel").aggregate({"revenue":"sum"})
emp_data.groupby("First Name").aggregate({"Salary":"count"},{"Salary":"mean"},{"Salary":"median"},{"Salary":"mode"})
```

```python
summary=pd.Series({
    "count":emp_data["Salary"].count(),
    "mean":emp_data["Salary"].mean(),
    "median":emp_data["Salary"].median(),
    "mode": emp_data["Salary"].mode().iloc[0]
})
```
